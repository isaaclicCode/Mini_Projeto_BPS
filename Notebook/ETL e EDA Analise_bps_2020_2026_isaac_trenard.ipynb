```python
#Carregamento de bibliotecas para concatenacao de arquivos csv
import pandas as pd
import unicodedata
import warnings
warnings.filterwarnings('ignore')

print('Carreganto de biblioteca com sucesso')
```

```python
# ============================================
# SPRINT 2: Importar os arquivos anuais
# ============================================

# Lista de anos e nomes dos arquivos
anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]
arquivos = [f'{ano}.csv' for ano in anos]

# Dicionário para armazenar os DataFrames
dataframes = {}

# Loop para carregar cada arquivo
for ano, arquivo in zip(anos, arquivos):
    try:
        dataframes[ano] = pd.read_csv(arquivo, sep=';', encoding='utf-8')
        print(f"✅ {ano} carregado: {dataframes[ano].shape[0]:,} registros")
    except FileNotFoundError:
        print(f"❌ Arquivo {arquivo} não encontrado!")
        exit()

print("\n📌 Todos os arquivos foram carregados com sucesso!")
```

```python
# ============================================
# ANÁLISE EXPLORATÓRIA DE DADOS (EDA) 
# ============================================

def explorar_dataframe(df, nome="DataFrame"):
    """
    Realiza uma análise exploratória básica em um DataFrame.
    Exibe: shape, colunas, dtypes, head, info, describe, nulos, duplicatas.
    """
    print("="*60)
    print(f"🔹 ANÁLISE EXPLORATÓRIA: {nome}")
    print("="*60)
    
    # 1. Shape (dimensões)
    print(f"\n📌 SHAPE: {df.shape[0]} linhas x {df.shape[1]} colunas")
    
    # 2. Colunas e tipos
    print("\n📌 COLUNAS E TIPOS:")
    print(df.dtypes.to_string())
    
    # 3. Primeiras linhas
    print("\n📌 HEAD (primeiras 5 linhas):")
    display(df.head())  # use print(df.head()) se não estiver no Jupyter
    
    # 4. Últimas linhas (opcional)
    print("\n📌 TAIL (últimas 5 linhas):")
    display(df.tail())
    
    # 5. Informações gerais (info)
    print("\n📌 INFO (resumo):")
    # Buffer para capturar a saída do info e exibir como string
    from io import StringIO
    buffer = StringIO()
    df.info(buf=buffer)
    print(buffer.getvalue())
    
    # 6. Estatísticas descritivas - numéricas
    print("\n📌 ESTATÍSTICAS DESCRITIVAS (numéricas):")
    display(df.describe())
    
    # 7. Estatísticas descritivas - categóricas
    cat_cols = df.select_dtypes(include=['object', 'category']).columns
    if len(cat_cols) > 0:
        print("\n📌 ESTATÍSTICAS DESCRITIVAS (categóricas):")
        display(df[cat_cols].describe())
    
    print("-"*60)



anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

for ano in anos:
    if ano in dataframes:
        explorar_dataframe(dataframes[ano], f"Base {ano} (original)")
    else:
        print(f"⚠️ Ano {ano} não encontrado.")
```

```python
# ============================================
# SPRINT 2: Verificar e padronizar os nomes das colunas
# ============================================

for ano, df in dataframes.items():
    print(f"{ano}: {df.shape[1]} colunas")

# 3. Verificar se todos têm as mesmas colunas
colunas_referencia = dataframes[2020].columns
todos_iguais = all(df.columns.equals(colunas_referencia) for df in dataframes.values())

if todos_iguais:
    print("\n✅ Todos os arquivos têm exatamente as mesmas colunas e ordem!")
else:
    print("\n⚠️ Há diferenças nas colunas. Serão padronizadas.")

# 4. Padronizar nomes das colunas
def padronizar_nomes_colunas(df):
    cols = df.columns.tolist()
    novos_nomes = []
    for col in cols:
        col = col.lower()
        col = col.replace(' ', '_').replace('-', '_')
        col = unicodedata.normalize('NFKD', col).encode('ASCII', 'ignore').decode('ASCII')
        novos_nomes.append(col)
    df.columns = novos_nomes
    return df

for ano, df in dataframes.items():
    dataframes[ano] = padronizar_nomes_colunas(df)

# 5. Verificar novamente após padronização
colunas_referencia = dataframes[2020].columns
todos_iguais_apos = all(df.columns.equals(colunas_referencia) for df in dataframes.values())

if todos_iguais_apos:
    print("\n✅ Após padronização, todos os DataFrames têm as mesmas colunas!")
else:
    print("\n❌ Ainda há diferenças. Verifique manualmente.")

# 6. Exibir os nomes padronizados
print("\n📌 Nomes padronizados das colunas (exemplo 2020):")
print(dataframes[2020].columns.tolist())
```

```python
# ============================================
# SPRINT 2: Padronizar campos de data (ano a ano)
# ============================================

print("🔹 Padronizando campos de data ")

anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

for ano in anos:
    df = dataframes[ano]
    
    # 1. Converter colunas para datetime
    df['compra'] = pd.to_datetime(df['compra'], format='%d/%m/%Y', errors='coerce')
    df['insercao'] = pd.to_datetime(df['insercao'], format='%d/%m/%Y', errors='coerce')
    
    # 2. Extrair componentes
    df['ano_compra'] = df['compra'].dt.year
    df['mes_compra'] = df['compra'].dt.month
    df['dia_compra'] = df['compra'].dt.day
    
    df['ano_insercao'] = df['insercao'].dt.year
    df['mes_insercao'] = df['insercao'].dt.month
    df['dia_insercao'] = df['insercao'].dt.day
    
    # 3. Estatísticas rápidas
    print(f"\n📌 {ano}:")
    print(f"   compra: {df['compra'].dtype} | período: {df['compra'].min()} a {df['compra'].max()}")
    print(f"   insercao: {df['insercao'].dtype} | nulos: {df['insercao'].isna().sum():,}")
    print(f"   ✅ Datas padronizadas")

print("\n✅ Padronização de datas concluída para todos os anos!")
```

```python
# ============================================
# SPRINT 2: Ajustar formatação de valores monetários e quantidades (ano a ano)
# ============================================

print("🔹 Ajustando formatação de valores monetários e quantidades...")

# Colunas a verificar
colunas_numericas = ['preco_unitario', 'preco_total', 'qtd_itens_comprados']
anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

for ano in anos:
    df = dataframes[ano]
    print(f"\n📌 {ano}:")
    
    for col in colunas_numericas:
        if col in df.columns:
            # 1. Verificar se há vírgula como separador decimal (se for object)
            if df[col].dtype == 'object':
                print(f"   ⚠️ {col}: é 'object' (pode ter vírgula). Corrigindo...")
                df[col] = df[col].astype(str).str.replace(',', '.')
                df[col] = pd.to_numeric(df[col], errors='coerce')
                print(f"      ✅ {col} convertido para {df[col].dtype}")
            else:
                print(f"   ✅ {col}: {df[col].dtype} (formato correto)")
            
            # 2. Verificar valores inconsistentes (negativos e zeros)
            negativos = (df[col] < 0).sum()
            zeros = (df[col] == 0).sum()
            if negativos > 0:
                print(f"      ⚠️ {negativos:,} valores negativos")
            if zeros > 0:
                print(f"      ℹ️ {zeros:,} valores zero")
    
    # 3. Estatísticas rápidas
    print(f"   📊 Resumo {ano}:")
    for col in colunas_numericas:
        if col in df.columns:
            print(f"      {col}: min={df[col].min():.2f}, max={df[col].max():.2f}, média={df[col].mean():.2f}")

print("\n✅ Ajuste de formatação concluído para todos os anos!")
```

```python
## ============================================
# SPRINT 2: Verificar nulos, vazios e inconsistentes 
# ============================================

print("🔹 Verificando valores nulos, vazios e inconsistentes \n")

anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]
resumo_anos = {}

for ano in anos:
    df = dataframes[ano]
    print(f"📌 {ano}:")
    
    # 1. Nulos por coluna (apenas com nulos > 0)
    nulos = df.isnull().sum()
    nulos = nulos[nulos > 0].sort_values(ascending=False)
    
    if len(nulos) > 0:
        print("   Nulos por coluna:")
        for col, qtd in nulos.items():
            print(f"      {col}: {qtd:,}")
    else:
        print("   ✅ Nenhum valor nulo encontrado.")
    
    # 2. Valores vazios em colunas de texto (object)
    colunas_texto = df.select_dtypes(include=['object']).columns
    vazios_ano = 0
    for col in colunas_texto:
        vazios = (df[col] == '').sum()
        if vazios > 0:
            print(f"   ⚠️ {col}: {vazios:,} valores vazios")
            vazios_ano += vazios
    if vazios_ano == 0:
        print("   ✅ Nenhum valor vazio encontrado.")
    
    # 3. Inconsistências numéricas
    colunas_numericas = ['preco_unitario', 'preco_total', 'qtd_itens_comprados']
    for col in colunas_numericas:
        if col in df.columns:
            negativos = (df[col] < 0).sum()
            zeros = (df[col] == 0).sum()
            if negativos > 0:
                print(f"   ⚠️ {col}: {negativos:,} valores negativos")
            if zeros > 0:
                print(f"   ℹ️ {col}: {zeros:,} valores zero")
            # Percentil 99 (outliers)
            q99 = df[col].quantile(0.99)
            if q99 > 0:
                print(f"   ℹ️ {col}: 99% dos valores estão abaixo de {q99:,.2f}")
    
    # 4. Inconsistências em datas
    if 'compra' in df.columns and 'insercao' in df.columns:
        data_invalida = (df['compra'] > df['insercao']).sum()
        if data_invalida > 0:
            print(f"   ⚠️ {data_invalida:,} registros com compra posterior à inserção")
        else:
            print("   ✅ Datas consistentes")
    
    # Armazenar resumo do ano
    resumo_anos[ano] = {
        'total_registros': len(df),
        'total_nulos': df.isnull().sum().sum(),
        'total_vazios': vazios_ano,
        'duplicados': df.duplicated().sum()
    }
    print("-" * 40)

# 5. Resumo geral
print("\n📌 RESUMO GERAL:")
total_registros = sum(r['total_registros'] for r in resumo_anos.values())
total_nulos = sum(r['total_nulos'] for r in resumo_anos.values())
total_vazios = sum(r['total_vazios'] for r in resumo_anos.values())
total_duplicados = sum(r['duplicados'] for r in resumo_anos.values())

print(f"   Total de registros (todos os anos): {total_registros:,}")
print(f"   Total de nulos: {total_nulos:,}")
print(f"   Total de valores vazios: {total_vazios:,}")
print(f"   Total de duplicados: {total_duplicados:,}")

print("\n✅ Verificação concluída para todos os anos!")
```

```python
# ============================================
# SPRINT 2 TRATAMENTO DE NULOS (coluna generico) 
# ============================================

def preencher_generico_3var(df):
    """Preenche nulos de 'generico' usando a moda por grupo de 3 variáveis."""
    nulos_iniciais = df['generico'].isna().sum()
    print(f"🔹 Nulos iniciais: {nulos_iniciais}")

    df = df.copy()
    df['chave'] = (
        df['descricao_catmat'].fillna('NaN').astype(str) + ' || ' +
        df['unidade_fornecimento'].fillna('NaN').astype(str) + ' || ' +
        df['fornecedor'].fillna('NaN').astype(str)
    )

    moda_por_grupo = {}
    for chave, grupo in df.groupby('chave'):
        valores = grupo['generico'].dropna()
        if len(valores) > 0:
            moda_por_grupo[chave] = valores.mode()[0]

    df['generico'] = df.apply(
        lambda row: moda_por_grupo.get(row['chave'], row['generico'])
        if pd.isna(row['generico']) else row['generico'],
        axis=1
    )
    df = df.drop(columns=['chave'])

    nulos_finais = df['generico'].isna().sum()
    print(f"✅ Nulos preenchidos (3 var): {nulos_iniciais - nulos_finais}")
    print(f"📌 Nulos restantes: {nulos_finais}")
    return df


def preencher_nao_informado_com_extras(df, colunas_extras=['capacidade', 'unidade_medida', 'fabricante'], limiar_confianca=0.8):
    """Substitui 'Não Informado' com base em colunas extras, se atingir o limiar."""
    mask_nao_informado = df['generico'] == 'Não Informado'
    if not mask_nao_informado.any():
        print("✅ Nenhum 'Não Informado' para processar.")
        return df

    print(f"🔹 'Não Informado' iniciais: {mask_nao_informado.sum()}")

    colunas_base = ['descricao_catmat', 'unidade_fornecimento', 'fornecedor']
    colunas_agrupamento = colunas_base + colunas_extras
    colunas_existentes = [col for col in colunas_agrupamento if col in df.columns]
    if len(colunas_existentes) < len(colunas_agrupamento):
        print(f"⚠️ Colunas não encontradas: {set(colunas_agrupamento) - set(colunas_existentes)}")
        colunas_agrupamento = colunas_existentes

    df['chave'] = df[colunas_agrupamento].fillna('NaN').astype(str).agg(' || '.join, axis=1)

    df_referencia = df[~mask_nao_informado].copy()
    proporcoes = df_referencia.groupby('chave')['generico'].value_counts(normalize=True).unstack().fillna(0)

    mapa_preenchimento = {}
    for chave in df.loc[mask_nao_informado, 'chave'].unique():
        if chave in proporcoes.index:
            prop_s = proporcoes.loc[chave].get('S', 0)
            prop_n = proporcoes.loc[chave].get('N', 0)
            if prop_s >= limiar_confianca:
                mapa_preenchimento[chave] = 'S'
            elif prop_n >= limiar_confianca:
                mapa_preenchimento[chave] = 'N'

    df.loc[mask_nao_informado, 'generico'] = df.loc[mask_nao_informado].apply(
        lambda row: mapa_preenchimento.get(row['chave'], row['generico']),
        axis=1
    )
    df = df.drop(columns=['chave'])

    preenchidos = (mask_nao_informado & (df['generico'] != 'Não Informado')).sum()
    print(f"✅ Preenchidos com extras: {preenchidos}")
    print(f"📌 Ainda 'Não Informado': {(df['generico'] == 'Não Informado').sum()}")
    return df


def tratar_coluna_generico(df):
    """Aplica as duas etapas de tratamento na coluna 'generico'."""
    df = preencher_generico_3var(df)
    df['generico'] = df['generico'].fillna('Não Informado')
    df = preencher_nao_informado_com_extras(df)
    return df


def analise_descritiva_generico(df, ano):
    """Exibe análise descritiva da coluna 'generico' para um ano específico."""
    contagem = df['generico'].value_counts()
    percentual = (contagem / len(df) * 100).round(2)
    resumo = pd.DataFrame({
        'Categoria': contagem.index,
        'Quantidade': contagem.values,
        'Percentual (%)': percentual.values
    })
    print("="*50)
    print(f"ANÁLISE DESCRITIVA - {ano} - COLUNA 'generico'")
    print("="*50)
    print(resumo.to_string(index=False))
    print(f"\nTotal de registros: {len(df):,}")
    print(f"Total preenchido (S + N): {contagem.get('S', 0) + contagem.get('N', 0):,}")
    print(f"Total 'Não Informado': {contagem.get('Não Informado', 0):,}")
    print("-"*50)

    # ============================================
# EXECUÇÃO PARA TODOS OS ANOS
# ============================================

anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

# Caso você já tenha os dataframes carregados em um dicionário chamado 'dataframes'
for ano in anos:
    if ano in dataframes:
        print(f"\n{'='*60}")
        print(f"🔹 PROCESSANDO ANO {ano}")
        print(f"{'='*60}")
        df = dataframes[ano].copy()          # cópia para não alterar o original
        df = tratar_coluna_generico(df)      # aplica o tratamento
        analise_descritiva_generico(df, ano) # exibe o resumo
        dataframes[ano] = df                 # (opcional) atualiza o dicionário
        # df.to_csv(f"{ano}_tratado.csv", index=False)  # (opcional) salvar
    else:
        print(f"⚠️ Ano {ano} não encontrado no dicionário 'dataframes'.")
```

```python
# ============================================
# SPRINT 2: TRATAMENTO COMPLETO DE DUPLICATAS
# ============================================

def tratar_duplicados(df, ano, colunas_parciais, colunas_contexto, top_n=3):
    """
    Executa o tratamento COMPLETO de duplicatas em um DataFrame de um único ano:
      FASE 1 → Avaliar duplicatas (completas + parciais)
      FASE 2 → Diagnosticar os maiores grupos com contexto
      FASE 3 → Classificar cada linha (unico / duplicata_real / compra_repetida)
      FASE 4 → Remover apenas 'duplicata_real', preservando compras repetidas legítimas
    
    Parâmetros:
        df: DataFrame do ano
        ano: int (para exibição)
        colunas_parciais: lista de colunas que formam a chave da duplicata
        colunas_contexto: colunas que diferenciam compras legítimas (data, instituição, etc.)
        top_n: quantos grupos mostrar no diagnóstico
    
    Retorna:
        df_limpo (DataFrame sem duplicatas reais) e um dict com o resumo.
    """
    
    # ============================================================
    # FASE 1: AVALIAR DUPLICATAS (COMPLETAS + PARCIAIS)
    # ============================================================
    print("="*60)
    print(f"📌 ANÁLISE DE DUPLICATAS - ANO {ano}")
    print("="*60)
    
    total_registros = len(df)
    print(f"Total de registros: {total_registros:,}")
    
    # 1.1 Duplicatas completas (todas as colunas)
    duplicados_completos = df.duplicated(keep=False)
    qtd_duplicados_completos = duplicados_completos.sum()
    qtd_linhas_unicas = total_registros - qtd_duplicados_completos
    
    print(f"\n🔹 DUPLICATAS COMPLETAS (todas as colunas):")
    print(f"   Registros duplicados: {qtd_duplicados_completos:,} ({qtd_duplicados_completos/total_registros*100:.2f}%)")
    print(f"   Registros únicos: {qtd_linhas_unicas:,}")
    
    if qtd_duplicados_completos > 0:
        amostra = df[duplicados_completos].head(5)
        print("\n   Exemplo de registros duplicados (primeiros 5):")
        print(amostra.to_string(index=False))
    
    # 1.2 Duplicatas parciais
    colunas_existentes = [c for c in colunas_parciais if c in df.columns]
    qtd_dup_parcial = 0
    if colunas_existentes:
        print(f"\n🔹 DUPLICATAS PARCIAIS (baseado em: {colunas_existentes}):")
        duplicados_parciais = df.duplicated(subset=colunas_existentes, keep=False)
        qtd_dup_parcial = duplicados_parciais.sum()
        print(f"   Registros com duplicatas parciais: {qtd_dup_parcial:,} ({qtd_dup_parcial/total_registros*100:.2f}%)")
        
        grupos = df.groupby(colunas_existentes).size()
        grupos_duplicados = grupos[grupos > 1]
        print(f"   Número de grupos duplicados: {len(grupos_duplicados):,}")
        
        if len(grupos_duplicados) > 0:
            print("\n   Maiores grupos de duplicatas parciais (top 5):")
            for idx, (keys, count) in enumerate(grupos_duplicados.nlargest(5).items()):
                keys_str = ' | '.join(str(k) for k in keys) if isinstance(keys, tuple) else str(keys)
                print(f"      {idx+1}. {keys_str[:120]} -> {count} registros")
    else:
        print("\n⚠️ Nenhuma das colunas parciais especificadas existe no DataFrame.")
    
    print(f"\n📊 RESUMO FASE 1:")
    print(f"   Total registros: {total_registros:,}")
    print(f"   Duplicatas completas: {qtd_duplicados_completos:,}")
    print(f"   Duplicatas parciais: {qtd_dup_parcial:,}")
    print("-"*60)
    
    # ============================================================
    # FASE 2: DIAGNÓSTICO DOS MAIORES GRUPOS (COM CONTEXTO)
    # ============================================================
    print("="*60)
    print(f"🔍 DIAGNÓSTICO DOS GRUPOS - ANO {ano}")
    print("="*60)
    
    ctx_existentes = [c for c in colunas_contexto if c in df.columns]
    
    if colunas_existentes:
        grupos = [(k, g) for k, g in df.groupby(colunas_existentes, dropna=False) if len(g) > 1]
        
        if not grupos:
            print(f"   ✅ Ano {ano}: nenhum grupo duplicado encontrado.")
        else:
            grupos.sort(key=lambda x: len(x[1]), reverse=True)
            print(f"   📌 Total de grupos duplicados: {len(grupos):,}")
            print(f"   📌 Mostrando os {top_n} maiores grupos:\n")
            
            for i, (chave, grupo) in enumerate(grupos[:top_n]):
                print("-"*60)
                print(f"🔹 GRUPO #{i+1} — {len(grupo)} registros")
                chave_str = ' | '.join(str(k)[:50] for k in chave) if isinstance(chave, tuple) else str(chave)[:100]
                print(f"   Chave: {chave_str}")
                
                cols_mostrar = colunas_existentes[:1] + ctx_existentes
                display_df = grupo[cols_mostrar].head(3).copy()
                for col in display_df.columns:
                    if display_df[col].dtype == 'object':
                        display_df[col] = display_df[col].astype(str).str.slice(0, 35)
                print(display_df.to_string(index=False))
                
                if len(grupo.drop_duplicates()) == 1:
                    print("   ⚠️  → DUPLICATA COMPLETA")
                else:
                    difs = [c for c in ctx_existentes if grupo[c].nunique(dropna=False) > 1]
                    if difs:
                        print(f"   💡 → COMPRA REPETIDA LEGÍTIMA (diferenças em: {difs})")
                    else:
                        print("   ⚠️  → PROVÁVEL DUPLICATA (contexto idêntico)")
            print("-"*60)
    
    # ============================================================
    # FASE 3: CLASSIFICAR CADA LINHA
    # ============================================================
    print("="*60)
    print(f"🏷️ CLASSIFICAÇÃO DE DUPLICATAS - ANO {ano}")
    print("="*60)
    
    df = df.copy()
    df['_classificacao'] = 'unico'
    
    if colunas_existentes:
        dup_mask = df.duplicated(subset=colunas_existentes, keep=False)
        
        for chave, grupo in df[dup_mask].groupby(colunas_existentes, dropna=False):
            idx = grupo.index
            
            # Se todas as colunas idênticas → duplicata real
            if len(df.loc[idx].drop_duplicates()) == 1:
                df.loc[idx, '_classificacao'] = 'duplicata_real'
                continue
            
            # Se há diferença em contexto → compra repetida
            dif_contexto = any(
                df.loc[idx, c].nunique(dropna=False) > 1
                for c in ctx_existentes if c in df.columns
            )
            df.loc[idx, '_classificacao'] = 'compra_repetida' if dif_contexto else 'duplicata_real'
    
    print("📊 Distribuição das classificações:")
    print(df['_classificacao'].value_counts().to_string())
    print("-"*60)
    
    # ============================================================
    # FASE 4: REMOVER APENAS DUPLICATAS REAIS
    # ============================================================
    print("="*60)
    print(f"🗑️ REMOÇÃO DE DUPLICATAS REAIS - ANO {ano}")
    print("="*60)
    
    antes = len(df)
    df_limpo = df[df['_classificacao'] != 'duplicata_real'].copy()
    df_limpo = df_limpo.drop(columns=['_classificacao'])
    removidos = antes - len(df_limpo)
    
    print(f"   Antes:     {antes:,} registros")
    print(f"   Depois:    {len(df_limpo):,} registros")
    print(f"   Removidos: {removidos:,} ({removidos/antes*100:.2f}%)")
    print(f"   Preservados como 'compra_repetida': {(df['_classificacao'] == 'compra_repetida').sum():,}")
    print("-"*60)
    
    # Resumo final do ano
    resumo = {
        'ano': ano,
        'total_registros': total_registros,
        'duplicados_completos': qtd_duplicados_completos,
        'duplicados_parciais': qtd_dup_parcial,
        'removidos': removidos,
        'total_final': len(df_limpo)
    }
    
    return df_limpo, resumo


# ============================================
# EXECUÇÃO (FORA DA FUNÇÃO — SEM INDENTAÇÃO)
# ============================================

anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

colunas_para_parciais = ['descricao_catmat', 'fornecedor', 'preco_total', 'qtd_itens_comprados']
colunas_contexto = ['compra', 'insercao', 'cnpj_instituicao', 'codigo_br',
                    'nome_instituicao', 'modalidade_compra']

resumos = []

for ano in anos:
    if ano in dataframes:
        print("\n" + "#"*60)
        print(f"# PROCESSANDO ANO {ano}")
        print("#"*60)
        df_limpo, resumo = tratar_duplicados(
            dataframes[ano], ano,
            colunas_parciais=colunas_para_parciais,
            colunas_contexto=colunas_contexto,
            top_n=3
        )
        dataframes[ano] = df_limpo
        resumos.append(resumo)
    else:
        print(f"⚠️ Ano {ano} não encontrado.")

# Resumo consolidado final
print("\n" + "="*60)
print("📋 RESUMO CONSOLIDADO — DUPLICATAS")
print("="*60)
df_resumo = pd.DataFrame(resumos)
print(df_resumo.to_string(index=False))
print("="*60)
```

```python
# ============================================
# CONCATENAÇÃO DAS BASES 2020-2026
# ============================================

def concatenar_bases(dataframes, anos, salvar_arquivo=None):
    """
    Concatena os DataFrames anuais tratados em um único DataFrame.
    A coluna 'ano_compra' (já presente em cada base) identifica a origem.
    
    Parâmetros:
        dataframes: dict {ano: DataFrame}
        anos: lista de anos a concatenar
        salvar_arquivo: (opcional) caminho para salvar o CSV resultante
    
    Retorna:
        df_consolidado (DataFrame único com todos os anos)
    """
    print("="*60)
    print("🔗 CONCATENAÇÃO DAS BASES 2020-2026")
    print("="*60)
    
    # FASE 1: Coletar apenas os anos existentes
    lista_dfs = []
    for ano in anos:
        if ano in dataframes:
            df = dataframes[ano]
            lista_dfs.append(df)
            print(f"   ✅ Ano {ano}: {len(df):,} registros | {df.shape[1]} colunas")
        else:
            print(f"   ⚠️ Ano {ano} não encontrado — pulando.")
    
    if not lista_dfs:
        print("❌ Nenhum DataFrame para concatenar.")
        return None
    
    # FASE 2: Concatenação (empilhamento vertical)
    print("\n📌 Concatenando...")
    df_consolidado = pd.concat(lista_dfs, ignore_index=True)
    
    # FASE 3: Resumo final
    print("\n" + "="*60)
    print("📊 RESULTADO DA CONCATENAÇÃO")
    print("="*60)
    print(f"   Total de registros: {len(df_consolidado):,}")
    print(f"   Total de colunas:   {df_consolidado.shape[1]}")
    print(f"\n   Distribuição por ano:")
    dist = df_consolidado['ano_compra'].value_counts().sort_index()
    for ano, qtd in dist.items():
        print(f"      {ano}: {qtd:,} registros")
    print("="*60)
    
    # FASE 4: Salvar (opcional)
    if salvar_arquivo:
        df_consolidado.to_csv(salvar_arquivo, index=False, encoding='utf-8')
        print(f"\n💾 Arquivo salvo em: {salvar_arquivo}")
    
    return df_consolidado

# ============================================
# EXECUÇÃO — CONCATENAR TODOS OS ANOS
# ============================================

anos = [2020, 2021, 2022, 2023, 2024, 2025, 2026]

df_consolidado = concatenar_bases(
    dataframes=dataframes,
    anos=anos,
    salvar_arquivo="dados_consolidados_BPS_2020_2026.csv"
)
```
